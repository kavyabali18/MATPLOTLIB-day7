```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Line Chart
x = [5, 10, 15, 20]
y = [12, 18, 28, 40]

plt.plot(x, y, color='green', marker='o')
plt.title("Sales Growth Line Chart")
plt.xlabel("Months")
plt.ylabel("Sales")

plt.show()
```

```python
# Bar Chart
days = ['Mon', 'Tue', 'Wed', 'Thu']
sales = [120, 180, 150, 200]
plt.bar(days, sales, color='orange')
plt.title("Weekly Sales Bar Chart")
plt.xlabel("Days")
plt.ylabel("Sales")

plt.show()
```

```python
# Histogram

data = [10, 12, 13, 15, 16, 18, 18, 20, 21,
        22, 23, 24, 25, 25, 26, 28, 30, 32]

plt.hist(data, bins=6, color='purple')

plt.title("Histogram Example")
plt.xlabel("Values")
plt.ylabel("Frequency")

plt.show()
```

```python
# Scatter Plot

x = [1, 2, 3, 4, 5, 6]
y = [10, 15, 13, 17, 22, 25]
plt.scatter(x, y, color='red')
plt.title("Scatter Plot Example")
plt.xlabel("X Values")
plt.ylabel("Y Values")

plt.show()
```

```python
# Pie Chart

brands = ['Apple', 'Samsung', 'Oppo', 'Vivo']
market = [35, 30, 20, 15]
plt.pie(market, labels=brands, autopct='%1.1f%%')
plt.title("Mobile Market Share")

plt.show()
```

```python
# Box Plot

data = [
    [5, 7, 8, 9, 10],
    [12, 14, 15, 16, 18],
    [20, 22, 23, 24, 25]
]
plt.boxplot(data)
plt.title("Box Plot Example")
plt.xlabel("Groups")
plt.ylabel("Values")

plt.show()
```

```python
# Heatmap

np.random.seed(1)
data = np.random.rand(8, 8)
plt.imshow(data, cmap='coolwarm')
plt.colorbar()
plt.title("Heatmap Example")
plt.show()
```
